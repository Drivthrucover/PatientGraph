```python
import pandas as pd
```

```python
def assign_center(row):
    s = row["symptom_code"]
    u = row["urgency"]
    loads = {
        "A": row["wait_load_A"],
        "B": row["wait_load_B"],
        "C": row["wait_load_C"]
    }

    if s == "cardiac" and row["specialty_match_A"]:
        return "Center_A"
    elif s == "respiratory" and row["specialty_match_B"]:
        return "Center_B"
    elif s == "injury" and row["specialty_match_C"]:
        return "Center_C"
    elif u >= 4:  # high urgency → send to least-loaded center
        return f"Center_{min(loads, key=loads.get)}"
    else:  # low urgency → still consider load balancing
        return f"Center_{min(loads, key=loads.get)}"
```

```python
def generate_labels(input_file="doctor_routing_raw.csv", output_file="doctor_routing_labeled.csv"):
    df = pd.read_csv("doctor_routing_raw.csv")
    df["assigned_center"] = df.apply(assign_center, axis=1)
    df.to_csv("labeled_patient_data.csv", index=False)
```

```python
if __name__ == "__main__":
    generate_labels()
```
